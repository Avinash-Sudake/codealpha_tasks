```python
# Import pandas library
import pandas as pd

# Load the CSV file into a DataFrame
df = pd.read_csv('/content/supermarket_Sales.csv')

# Display the first 5 rows of the dataset
df.head()
```

```python
df.info()
```

```python
# Display statistical summary of numerical columns
df.describe()
```

```python
# Check for missing values in each column
df.isnull().sum()
```

```python
# Check the number of duplicate rows
df.duplicated().sum()
```

```python
# Calculate total sales by city
city_sales = df.groupby('City')['Total'].sum()

# Display total sales by city
print(city_sales)
```

```python
# Import matplotlib library
import matplotlib.pyplot as plt

# Calculate total sales by city
city_sales = df.groupby('City')['Total'].sum()

# Create a bar chart
plt.figure(figsize=(6,4))
city_sales.plot(kind='bar')

# Add chart title and labels
plt.title('Total Sales by City')
plt.xlabel('City')
plt.ylabel('Total Sales')

# Display the chart
plt.show()
```

```python
# Calculate total sales by product line
product_sales = df.groupby('Product line')['Total'].sum()

# Display total sales by product line
print(product_sales)

# Import matplotlib
import matplotlib.pyplot as plt

# Create a bar chart
plt.figure(figsize=(10,5))
product_sales.plot(kind='bar')

# Add chart title and labels
plt.title('Total Sales by Product Line')
plt.xlabel('Product Line')
plt.ylabel('Total Sales')

# Display the chart
plt.show()
```

```python
# Calculate total sales by gender
gender_sales = df.groupby('Gender')['Total'].sum()

# Display total sales by gender
print(gender_sales)

# Import matplotlib
import matplotlib.pyplot as plt

# Create a pie chart
plt.figure(figsize=(6,6))
gender_sales.plot(kind='pie', autopct='%1.1f%%')

# Add chart title
plt.title('Sales Distribution by Gender')

# Remove y-axis label
plt.ylabel('')

# Display the chart
plt.show()
```

```python
D# Calculate total sales by payment method
payment_sales = df.groupby('Payment')['Total'].sum()

# Display total sales by payment method
print(payment_sales)

# Import matplotlib
import matplotlib.pyplot as plt

# Create a bar chart
plt.figure(figsize=(6,4))
payment_sales.plot(kind='bar')

# Add chart title and labels
plt.title('Total Sales by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Total Sales')

# Display the chart
plt.show()
```

```python
# Convert Date column to datetime format
df['Date'] = pd.to_datetime(df['Date'])

# Create a Month column
df['Month'] = df['Date'].dt.month_name()

# Calculate total sales by month
monthly_sales = df.groupby('Month')['Total'].sum()

# Display monthly sales
print(monthly_sales)

# Import matplotlib
import matplotlib.pyplot as plt

# Create a line chart
plt.figure(figsize=(8,5))
monthly_sales.plot(kind='line', marker='o')

# Add chart title and labels
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')

# Display the chart
plt.show()
```
